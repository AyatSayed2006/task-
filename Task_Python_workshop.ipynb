## Question 1 — Grocery Receipt
Write a program that asks the user for:
1. The name of an item
2. The price of one unit
3. The quantity being bought

Then calculate the total cost and print a simple receipt.

**Example:**
```
Enter item name: Notebook
Enter price per unit: 3.5
Enter quantity: 4
----- RECEIPT -----
Item: Notebook
Total: 14.0 NIS
```

````python
# Write your Question 1 solution here
item_name = input("Enter the item name: ")
price = float(input("Enter the price per unit: "))
quantity = int(input("Enter the quantity: "))
total_cost = price * quantity
print("----- RECEIPT -----")
print(f"Item: {item_name}")
print(f"Total: {total_cost}")
````

## Question 2 — Movie Ticket Pricing

Ask the user for their age and print which ticket type they get and its price.


## Question 3 — Weekly Study Tracker
Using a loop, ask the user to enter how many hours they studied for **each of the 5 weekdays** (Sunday–Thursday).

While collecting the values:
- Keep a running total of hours.

After the loop ends:
- Calculate the **average** study hours per day.
- If the average is 3 hours or more, print `"Great study habits!"`
- Otherwise, print `"Try to study a bit more next week."`

**Example:**
```
Enter hours studied on day 1: 2
Enter hours studied on day 2: 3
Enter hours studied on day 3: 4
Enter hours studied on day 4: 1
Enter hours studied on day 5: 5
Average hours per day: 3.0
Great study habits!
```

````python
# Write your Question 3 solution here
total_hours = 0
for day in range (1,6):
    hours = float(input(f"Enter the number of hours studied on day {day}: "))
    total_hours += hours

average = total_hours / 5
print(f"Average hours per day: {average}")
if average >= 3:
    print("Great study habits!")
else:
    print("Try to study a bit more next week.")
    
````
